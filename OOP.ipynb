```python
class circle:
    pi = 3.14
    def __init__(self,radius):
        self.radiusC = radius
    def get_circumference(self):
        circum= 2*self.pi*self.radiusC
        return circum
circle_1 = circle(7)
print(circle_1.get_circumference())
```
