```python
#functions
def my_car(type_of_car):
    print(f"My type of car is {type_of_car}" )
my_car("Toyota")
```

```python
#positional argument
def my_cars(car1, car2):
    print(f" I have a {car1} and a {car2} in my garage")
my_cars("Toyota","Tesla")
```

```python
#keyword argument in python
def my_cars(car2, car1 = "Toyota"):
    print(f" I have a {car1} and a {car2} in my garage")
my_cars("Toyota","Hilux")
    
```

```python
#using *args and **kwargs
def my_cars(*args):
    for car in args:
        print(" i have a", car )
my_cars("Toyota","Hilax","Tesla","Range Rover")
```

```python
def my_cars(**kwargs):
    
    for key,value in kwargs.items():
        print(f" {key}: {value}")
my_cars(age = 20, miles = 57, country = "usa")

```

```python
#Recursion
def factorial(n):
    if n ==0:
        return 1
    return n* factorial(n-1)
print(factorial(5))
```

```python
def countdown(n):
    if n ==0:
        print("Go")
        return 
    print(n)
    countdown(n-1)
countdown(3)
```
