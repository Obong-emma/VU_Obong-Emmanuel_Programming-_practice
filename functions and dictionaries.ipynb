```python
def prime_number(number):
    is_prime = True
    for i in range(2,number):
        if number % i ==0:
            is_prime = False
    if is_prime == True:
        print("Its a prime number")
    else:
        print("Not a prime number")
n = int(input("Enter a number:\n"))
prime_number(n)
```
